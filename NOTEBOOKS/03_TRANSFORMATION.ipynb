```python
import pandas as pd

df = pd.read_csv("../DATA/RAW/dirty_cafe_sales.csv")

# On applique à nouveau les étapes de nettoyage
df = df.drop_duplicates()
df["Transaction Date"] = pd.to_datetime(df["Transaction Date"], errors="coerce")
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Price Per Unit"] = pd.to_numeric(df["Price Per Unit"], errors="coerce")
df = df.dropna(subset=["Transaction Date", "Item", "Quantity", "Price Per Unit"])
df = df[(df["Quantity"] > 0) & (df["Price Per Unit"] > 0)]
df["Payment Method"] = df["Payment Method"].fillna("Unknown")
```

Le prix total est recalculé afin de corriger les incohérences présentes dans le dataset initial.

```python
df["Total_Price"] = df["Quantity"] * df["Price Per Unit"]
```

```python
df["Payment Method"] = df["Payment Method"].str.lower().str.strip()
df["Item"] = df["Item"].str.lower().str.strip()
```

```python
print(df.columns)
```

```python
df["Year"] = df["Transaction Date"].dt.year
df["Month"] = df["Transaction Date"].dt.month
```
